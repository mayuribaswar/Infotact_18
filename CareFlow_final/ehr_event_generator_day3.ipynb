```python
"""
============================================================
CareFlow - Hospital EHR Event Generator
Day 3: Add Case ID and Patient ID

Objective:
Generate a synthetic hospital event log with unique Case IDs
and Patient IDs for each hospital visit.
============================================================
"""

import pandas as pd
import random
from datetime import datetime, timedelta

# ============================================================
# Configuration
# ============================================================

NUM_PATIENTS = 1000
START_DATE = datetime(2026, 1, 1, 8, 0)

random.seed(42)

# ============================================================
# Hospital Workflow
# ============================================================

activities = [
    "Patient Arrival",
    "Registration",
    "Waiting",
    "Initial Assessment",
    "Doctor Consultation",
    "Lab Test Ordered",
    "Sample Collection",
    "Lab Processing",
    "Lab Result Available",
    "Diagnosis",
    "Treatment Started",
    "Medicine Dispensed",
    "Monitoring",
    "Billing",
    "Payment",
    "Discharged"
]

# ============================================================
# Department Mapping
# ============================================================

departments = {
    "Patient Arrival": "Reception",
    "Registration": "Reception",
    "Waiting": "Reception",
    "Initial Assessment": "Nursing",
    "Doctor Consultation": "Clinical",
    "Lab Test Ordered": "Laboratory",
    "Sample Collection": "Laboratory",
    "Lab Processing": "Laboratory",
    "Lab Result Available": "Laboratory",
    "Diagnosis": "Clinical",
    "Treatment Started": "Treatment",
    "Medicine Dispensed": "Pharmacy",
    "Monitoring": "Ward",
    "Billing": "Billing",
    "Payment": "Billing",
    "Discharged": "Reception"
}

# ============================================================
# Helper Functions
# ============================================================

def generate_case_id(case_number):
    """Generate unique Case ID."""
    return f"CASE-{case_number:05d}"


def generate_patient_id(patient_number):
    """Generate unique Patient ID."""
    return f"PAT-{patient_number:05d}"


def generate_doctor_id():
    """Generate Doctor ID."""
    return f"DOC{random.randint(101,120)}"


def generate_nurse_id():
    """Generate Nurse ID."""
    return f"NUR{random.randint(201,220)}"


def generate_severity():
    """Generate Severity."""
    return random.choice([
        "Low",
        "Medium",
        "High",
        "Critical"
    ])


# ============================================================
# Generate Event Log
# ============================================================

rows = []

for i in range(1, NUM_PATIENTS + 1):

    case_id = generate_case_id(i)
    patient_id = generate_patient_id(i)

    doctor_id = generate_doctor_id()
    nurse_id = generate_nurse_id()

    severity = generate_severity()

    current_time = START_DATE + timedelta(
        minutes=random.randint(0, 50000)
    )

    for activity in activities:

        current_time += timedelta(
            minutes=random.randint(5, 30)
        )

        rows.append({

            "Case_ID": case_id,

            "Patient_ID": patient_id,

            "Activity": activity,

            "Department": departments[activity],

            "Timestamp": current_time.strftime("%Y-%m-%d %H:%M:%S"),

            "Doctor_ID": doctor_id,

            "Nurse_ID": nurse_id,

            "Severity": severity,

            "Waiting_Time_Minutes": random.randint(5, 30),

            "Status": "Completed"

        })

# ============================================================
# Create DataFrame
# ============================================================

df = pd.DataFrame(rows)

# ============================================================
# Validation
# ============================================================

print("\n================= VALIDATION =================\n")

print(f"Unique Patients : {df['Patient_ID'].nunique()}")
print(f"Unique Cases    : {df['Case_ID'].nunique()}")
print(f"Total Records   : {len(df)}")

print("\nEvents Per Case (First 5 Cases)\n")

print(df.groupby("Case_ID").size().head())

case_validation = (
    df.groupby("Case_ID")["Patient_ID"]
      .nunique()
)

if case_validation.max() == 1:
    print("\n✓ Every Case_ID belongs to exactly one Patient_ID.")
else:
    print("\n✗ Invalid Case_ID mapping detected.")

patient_validation = (
    df.groupby("Patient_ID")["Case_ID"]
      .nunique()
)

print(f"\nMaximum Cases per Patient : {patient_validation.max()}")

print("\n==============================================\n")

# ============================================================
# Save CSV
# ============================================================

OUTPUT_FILE = "hospital_event_log_raw.csv"

df.to_csv(
    OUTPUT_FILE,
    index=False
)

# ============================================================
# Final Output
# ============================================================

print("=" * 60)
print("CareFlow Hospital Event Log Generated Successfully")
print("=" * 60)

print(f"Patients            : {NUM_PATIENTS}")
print(f"Unique Patient IDs  : {df['Patient_ID'].nunique()}")
print(f"Unique Case IDs     : {df['Case_ID'].nunique()}")
print(f"Activities          : {len(activities)}")
print(f"Total Records       : {len(df)}")
print(f"CSV File            : {OUTPUT_FILE}")

print("=" * 60)

print("\nSample Records\n")

display(df.head(10))
```
