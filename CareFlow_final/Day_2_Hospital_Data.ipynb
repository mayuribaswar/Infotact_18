```python
# Day 2: Create Basic EHR Event Generator:
# Objective: Generate a basic hospital event log for multiple patients based on the defined 
# hospital workflow.

import pandas as pd
import random
from datetime import datetime, timedelta

# -------------------------------------
# Configuration
# -------------------------------------

NUM_PATIENTS = 1000
START_DATE = datetime(2026, 1, 1, 8, 0)

random.seed(42)

# -------------------------------------
# Hospital Workflow
# -------------------------------------

activities = [
    "Patient Arrival",
    "Registration",
    "Waiting",
    "Initial Assessment",
    "Doctor Consultation",
    "Lab Test Ordered",
    "Sample Collection",
    "Lab Processing",
    "Lab Result Available",
    "Diagnosis",
    "Treatment Started",
    "Medicine Dispensed",
    "Monitoring",
    "Billing",
    "Payment",
    "Discharged"
]

# -------------------------------------
# Department Mapping
# -------------------------------------

departments = {
    "Patient Arrival": "Reception",
    "Registration": "Reception",
    "Waiting": "Reception",
    "Initial Assessment": "Nursing",
    "Doctor Consultation": "Clinical",
    "Lab Test Ordered": "Laboratory",
    "Sample Collection": "Laboratory",
    "Lab Processing": "Laboratory",
    "Lab Result Available": "Laboratory",
    "Diagnosis": "Clinical",
    "Treatment Started": "Treatment",
    "Medicine Dispensed": "Pharmacy",
    "Monitoring": "Ward",
    "Billing": "Billing",
    "Payment": "Billing",
    "Discharged": "Reception"
}

# -------------------------------------
# Generate Event Log
# -------------------------------------

rows = []

for i in range(1, NUM_PATIENTS + 1):

    case_id = f"C{i:05d}"
    patient_id = f"P{i:05d}"

    current_time = START_DATE + timedelta(
        minutes=random.randint(0, 50000)
    )

    doctor_id = f"DOC{random.randint(101,120)}"
    nurse_id = f"NUR{random.randint(201,220)}"

    severity = random.choice([
        "Low",
        "Medium",
        "High",
        "Critical"
    ])

    for activity in activities:

        current_time += timedelta(
            minutes=random.randint(5,30)
        )

        rows.append({

            "Case_ID": case_id,

            "Patient_ID": patient_id,

            "Activity": activity,

            "Department": departments[activity],

            "Timestamp": current_time.strftime("%Y-%m-%d %H:%M:%S"),

            "Doctor_ID": doctor_id,

            "Nurse_ID": nurse_id,

            "Severity": severity,

            "Waiting_Time_Minutes": random.randint(5,30),

            "Status": "Completed"

        })

# -------------------------------------
# Create DataFrame
# -------------------------------------

df = pd.DataFrame(rows)

# -------------------------------------
# Save CSV
# -------------------------------------

df.to_csv(
    "hospital_event_log_raw.csv",
    index=False
)

# -------------------------------------
# Output
# -------------------------------------

print("=" * 60)
print("Hospital EHR Event Log Generated Successfully")
print("=" * 60)
print(f"Patients       : {NUM_PATIENTS}")
print(f"Total Records  : {len(df)}")
print(f"Activities     : {len(activities)}")
print("=" * 60)

print(df.head())
```
