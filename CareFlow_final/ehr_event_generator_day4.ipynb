```python
"""
CareFlow - Hospital EHR Event Generator
Day 4: Add timestamps and event fields
"""
import pandas as pd
import random
from datetime import datetime, timedelta

NUM_PATIENTS = 1000
START_DATE = datetime(2026,1,1,8,0)
random.seed(42)

activities = [
    "Patient Arrival","Registration","Waiting","Initial Assessment",
    "Doctor Consultation","Lab Test Ordered","Sample Collection",
    "Lab Processing","Lab Result Available","Diagnosis",
    "Treatment Started","Medicine Dispensed","Monitoring",
    "Billing","Payment","Discharged"
]

departments = {
    "Patient Arrival":"Reception","Registration":"Reception","Waiting":"Reception",
    "Initial Assessment":"Nursing","Doctor Consultation":"Clinical",
    "Lab Test Ordered":"Laboratory","Sample Collection":"Laboratory",
    "Lab Processing":"Laboratory","Lab Result Available":"Laboratory",
    "Diagnosis":"Clinical","Treatment Started":"Treatment",
    "Medicine Dispensed":"Pharmacy","Monitoring":"Ward",
    "Billing":"Billing","Payment":"Billing","Discharged":"Reception"
}

activity_duration = {
    "Patient Arrival":2,"Registration":10,"Waiting":20,
    "Initial Assessment":15,"Doctor Consultation":20,
    "Lab Test Ordered":5,"Sample Collection":10,
    "Lab Processing":40,"Lab Result Available":5,
    "Diagnosis":10,"Treatment Started":30,
    "Medicine Dispensed":10,"Monitoring":60,
    "Billing":15,"Payment":5,"Discharged":5
}

def generate_case_id(n): return f"CASE-{n:05d}"
def generate_patient_id(n): return f"PAT-{n:05d}"
def doctor_id(): return f"DOC{random.randint(101,120)}"
def nurse_id(): return f"NUR{random.randint(201,220)}"

rows=[]
event_counter=1

for i in range(1,NUM_PATIENTS+1):
    case_id=generate_case_id(i)
    patient_id=generate_patient_id(i)
    current_time=START_DATE+timedelta(minutes=random.randint(0,50000))
    doc=doctor_id()
    nurse=nurse_id()
    severity=random.choice(["Low","Medium","High","Critical"])
    visit_type=random.choice(["OPD","IPD","Emergency"])
    room=f"R-{random.randint(101,120)}"
    bed=f"B-{random.randint(1,40):02d}"
    diagnosis=f"D{random.randint(100,150)}"
    treatment=random.choice(["Medication","Observation","Physiotherapy","Surgery"])

    for seq,activity in enumerate(activities,start=1):
        current_time += timedelta(minutes=activity_duration[activity])
        rows.append({
            "Event_ID":f"EVT{event_counter:07d}",
            "Event_Sequence":seq,
            "Case_ID":case_id,
            "Patient_ID":patient_id,
            "Visit_Type":visit_type,
            "Activity":activity,
            "Department":departments[activity],
            "Timestamp":current_time.strftime("%Y-%m-%d %H:%M:%S"),
            "Doctor_ID":doc,
            "Nurse_ID":nurse,
            "Room_Number":room,
            "Bed_Number":bed,
            "Diagnosis_Code":diagnosis,
            "Treatment_Type":treatment,
            "Severity":severity,
            "Waiting_Time_Minutes":random.randint(5,30),
            "Cost":random.randint(200,5000),
            "Status":"Completed"
        })
        event_counter+=1

df=pd.DataFrame(rows)
display(df)

print("Unique Patients:",df["Patient_ID"].nunique())
print("Unique Cases:",df["Case_ID"].nunique())
print("Total Records:",len(df))

df.to_csv("hospital_event_log_day4.csv",index=False)
print("Saved hospital_event_log_day4.csv")
```
