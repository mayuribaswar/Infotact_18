```python
import pandas as pd

# 1. Load raw CSV
df = pd.read_csv("hospital_event_log_raw.csv")

print("Original shape:", df.shape)
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
df["Timestamp"] = pd.to_datetime(df["Timestamp"], errors="coerce")
```

```python
df["Waiting_Time_Minutes"] = pd.to_numeric(
    df["Waiting_Time_Minutes"], errors="coerce"
)

df["Cost"] = pd.to_numeric(
    df["Cost"], errors="coerce"
)
```

```python
print("Negative waiting times:",
      (df["Waiting_Time_Minutes"] < 0).sum())

print("Negative costs:",
      (df["Cost"] < 0).sum())
```

```python
df.to_csv("hospital_event_log_cleaned.csv", index=False)

print("Cleaned shape:", df.shape)
print("Cleaned CSV created successfully!")
```
